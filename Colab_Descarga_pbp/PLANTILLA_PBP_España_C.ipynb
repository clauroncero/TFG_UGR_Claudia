importante!!!
recopilacion de acciones -> hay que recorrer todos los cuartos del partido (Q1, Q2,..) + prorrogas (OT1,...) -> se almacena todas las acciones en una lista

```python
try:


    # ID DEL PARTIDO

    game_id = xxxxxx

    # URL del Play-by-Play
    url = f"https://gdap.fiba.basketball/v2/games/xxxxx/play-by-play"

    # Subscription Key de FIBA
    hdr = {
        'Cache-Control': 'no-cache',
        'Ocp-Apim-Subscription-Key': 'XXXXXXXXXXXXXXXXXXXX'
    }

    # Crear petición
    req = urllib.request.Request(url, headers=hdr)
    req.get_method = lambda: 'GET'

    # Ejecutar petición
    response = urllib.request.urlopen(req)
    print("Código de respuesta:", response.getcode())

    # Leer respuesta
    data = response.read()
    text = data.decode('utf-8')
    json_obj = json.loads(text)

    # Lista con todas las acciones
    all_events = []

    # Recorrer todos los cuartos y prórrogas
    for quarter in ["Q1", "Q2", "Q3", "Q4", "OT1", "OT2", "OT3"]:
        if quarter in json_obj:
            for play in json_obj[quarter]["Items"]:
                play["Quarter"] = quarter
                all_events.append(play)

    # Crear DataFrame
    df = pd.DataFrame(all_events)

    # Guardar en Google Drive
    df.to_csv(
        f"/content/drive/MyDrive/FIBA_PlayByPlay/pbp_xxxxx.csv",
        index=False,
        encoding="utf-8-sig"
    )

    print(f"\nArchivo guardado correctamente: pbp_xxxxx.csv")
    print(f"Número de acciones: {len(df)}")

    print("\nPrimeras filas:")
    display(df.head())

except Exception as e:
    print("Error:", e)
```
